```python
# Sales Data Analysis - Exploration

import pandas as pd
import matplotlib.pyplot as plt

# Create sample sales dataset
data = {
    "Date": [
        "2026-01-05", "2026-01-10", "2026-02-03",
        "2026-02-15", "2026-03-08", "2026-03-20",
        "2026-04-04", "2026-04-18", "2026-05-10",
        "2026-05-22"
    ],
    "Product": [
        "Laptop", "Mobile", "Laptop", "Tablet",
        "Mobile", "Laptop", "Tablet", "Mobile",
        "Laptop", "Tablet"
    ],
    "Quantity": [2, 5, 3, 4, 6, 2, 5, 3, 4, 6],
    "Sales": [100000, 75000, 150000, 80000, 90000,
             110000, 100000, 45000, 200000, 120000],
    "Region": [
        "North", "South", "East", "West", "North",
        "South", "East", "West", "North", "South"
    ]
}

df = pd.DataFrame(data)

# Display dataset
print("SALES DATASET")
display(df)

# Dataset information
print("\nDATASET INFORMATION")
df.info()

# Statistical summary
print("\nSTATISTICAL SUMMARY")
display(df.describe())

# Check missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# Check duplicate records
print("\nDUPLICATE RECORDS")
print(df.duplicated().sum())

# Total and average sales
print("\nTOTAL SALES:", df["Sales"].sum())
print("AVERAGE SALES:", df["Sales"].mean())

# Top selling products
print("\nTOP SELLING PRODUCTS")
display(
    df.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

# Monthly sales
df["Date"] = pd.to_datetime(df["Date"])

monthly_sales = (
    df.groupby(df["Date"].dt.month)["Sales"].sum()
)

print("\nMONTHLY SALES")
display(monthly_sales)

# Product sales chart
df.groupby("Product")["Sales"].sum().plot(
    kind="bar", title="Sales by Product"
)
plt.ylabel("Sales")
plt.tight_layout()
plt.show()

# Monthly sales chart
monthly_sales.plot(
    kind="line", marker="o", title="Monthly Sales Trend"
)
plt.ylabel("Sales")
plt.xlabel("Month")
plt.tight_layout()
plt.show()

# Sales distribution
df.groupby("Product")["Sales"].sum().plot(
    kind="pie", autopct="%1.1f%%",
    title="Sales Distribution"
)
plt.ylabel("")
plt.show()

print("\nExploratory Data Analysis Completed!")
```
